PART 1- Mean, median, mode

```python
salaries = [    
 18, 20, 22, 22, 24,   
 25, 25, 26, 28, 30,    
 32, 35, 38, 42, 150]
```

```python

def calculate_mean(data):
    return sum(data)/len(data)
print("Mean:", calculate_mean(salaries))

def calculate_median(data):
    data.sort()
    n = len(salaries)
    if n % 2 == 0:
      return (salaries[n//2 - 1] + salaries[n//2]) / 2
    else:
      return salaries[n//2]
print("Median:", calculate_median(salaries))

def calculate_mode(data):
        max_count = 0
        modes = []
        for i in data:
          count = data.count(i)
        if count > max_count:
            max_count = count
            modes = [i]
        elif count == max_count and i not in modes:
            modes.append(i)
        return modes
print("Mode:", calculate_mode(salaries))
```

```python
data_1 = [12, 15, 15, 18, 20, 22, 25]
data_2 = [5, 5, 10, 10, 15, 20,0]
data_3 = [7, 14, 21, 28, 35, 0, 0]
data_4 = [100, 100, 100, 100,0,0,0]
data_5 = [1, 2, 3, 4, 5, 100,0]


def mean(data):
    total = 0
    for value in data:
        total += value

    return total / len(data)

def median(data):
    numbers = sorted(data)
    n = len(data)
    mid = n // 2

    if n % 2 == 0:
        return (data[mid - 1] + data[mid]) / 2
    else:
        return data[mid]

def mode(data):
    return max(set(data), key=data.count)

datasets = {
    "Dataset 1": data_1,
    "Dataset 2": data_2,
    "Dataset 3": data_3,
    "Dataset 4": data_4,
    "Dataset 5": data_5
}
print(f"{'Dataset':<12} {'Mean':>10} "
      f"{'Median':>10} {'Mode':>15}")
print("-" * 50)

for name, data in datasets.items():
    mean = calculate_mean(data)
    median = calculate_median(data)
    mode = calculate_mode(data)

    mode_text = str(mode) if mode else "No mode"

    print(f"{name:<12} {mean:>10.2f} "
          f"{median:>10.2f} {mode_text:>15}")
```

PART 2- variance, standard deviation


```python
import numpy as np
scores = [45, 50, 52, 55, 58, 60, 65, 75]

mean = sum(scores)/len(scores)
print("mean:", mean)
print("---------------------------------------------------------")
def population_variance(data):
    mean = sum(data) / len(data)
    return sum((x - mean) ** 2 for x in data) / len(data)
print("population variance:", population_variance(scores))
print("---------------------------------------------------------")
def sample_variance(data):
    mean = sum(data) / len(data)
    return sum((x - mean) ** 2 for x in data) / (len(data)-1)
print("sample variance:", sample_variance(scores))
print("---------------------------------------------------------")
def popu_std(data):
    mean = sum(data) / len(data)
    variance = sum((x - mean) ** 2 for x in data) / len(data)
    return variance ** 0.5
print("population std:", popu_std(scores))
print("---------------------------------------------------------")
def sample_std(data):
    mean = sum(data) / len(data)
    samp_var = sum((x - mean) ** 2 for x in data) / (len(data)-1)
    return samp_var ** 0.5
print("sample std:", sample_std(scores))
```

```python
import seaborn as sns
import numpy as np
import pandas as pd

tips = sns.load_dataset("tips")
col = ["total_bill", "tip"]
results = {}

for x in col:
    mean = tips[x].mean()
    variance = tips[x].var(ddof=0)
    std_dev = tips[x].std(ddof=0)
    cv = (std_dev/mean)*100

results[x] = {
        "Mean": mean,
        "Population Variance": variance,
        "Population Standard Deviation": std_dev,
        "CV (%)": cv
    }
df_results = pd.DataFrame(results).T

df_results.round(4)
```

```python
largest_bills = tips.nlargest(5, "total_bill")[["total_bill"]]
print("Five largest total_bill values:", largest_bills)

og = tips["total_bill"]

# Temporarily exclude the five largest observations
filtered = og.drop(largest_bills.index)


def calculate_stats(data):
    return {
        "Mean": data.mean(),
        "Population Variance": data.var(ddof=0),
        "Population Standard Deviation": data.std(ddof=0)
    }

comparison = pd.DataFrame({
    "Original Data": calculate_stats(og),
    "After Exclusion": calculate_stats(filtered)
})

comparison["Change"] = (
    comparison["After Exclusion"] - comparison["Original Data"]
)

print("\nStatistical comparison:")
print(comparison.round(4))
```

```python
import math

data_1 = [10, 20, 30, 40, 50]
data_2 = [20, 20, 20, 20, 20]
data_3 = [5, 10, 15, 20, 100]
data_4 = [12, 14, 16, 18, 20, 22]

def c_mean(data):
    return sum(data) / len(data)

def p_var(data):
    mean = calculate_mean(data)
    squared_deviations = [(x - mean) ** 2 for x in data]
    return sum(squared_deviations) / len(data)

def s_var(data):
    mean = calculate_mean(data)
    squared_deviations = [(x - mean) ** 2 for x in data]
    return sum(squared_deviations) / (len(data)-1)

def p_std(data):
    mean = sum(data) / len(data)
    variance = sum((x - mean) ** 2 for x in data) / len(data)
    return variance ** 0.5

def s_std(data):
    mean = sum(data) / len(data)
    samp_var = sum((x - mean) ** 2 for x in data) / (len(data)-1)
    return samp_var ** 0.5

ds = {
    "data_1": data_1,
    "data_2": data_2,
    "data_3": data_3,
    "data_4": data_4
}

results = {}

for name, data in ds.items():
    results[name] = {
        "Count": len(data),
        "Mean": c_mean(data),
        "Population Variance": p_var(data),
        "Sample Variance": s_var(data),
        "Population Std Dev": p_std(data),
        "Sample Std Dev": s_std(data)
    }
df1=pd.DataFrame(results)
df1
```

```python
sales = [12, 14, 15, 15, 16, 17, 18,
         19, 20, 21, 22, 24, 25, 120]
mean = statistics.mean(sales)
median = statistics.median(sales)
popu_variance = statistics.pvariance(sales)
popu_std_dev = statistics.pstdev(sales)

sorted_sales = sorted(sales)
n = len(sorted_sales)

lower_half = sorted_sales[:n // 2]
upper_half = sorted_sales[n // 2:]

Q1 = statistics.median(lower_half)
Q3 = statistics.median(upper_half)

IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify potential outliers
outliers = [x for x in sales
            if x < lower_bound or x > upper_bound]

df4 = pd.DataFrame({
    'Statistic': [
        'Mean',
        'Median',
        'Population Variance',
        'Population Standard Deviation',
        'First Quartile (Q1)',
        'Third Quartile (Q3)',
        'Interquartile Range (IQR)',
        'Lower Bound',
        'Upper Bound',
        'Flagged Observations'
    ],
    'Value': [
        mean,
        median,
        popu_variance,
        popu_std_dev,
        Q1,
        Q3,
        IQR,
        lower_bound,
        upper_bound,
        str(outliers)
    ]
})

df4
```
