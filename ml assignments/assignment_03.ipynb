PART 1- Mean, median, mode

```python
salaries = [    
 18, 20, 22, 22, 24,   
 25, 25, 26, 28, 30,    
 32, 35, 38, 42, 150]
```

```python

def calculate_mean(data):
    return sum(data)/len(data)
print("Mean:", calculate_mean(salaries))

def calculate_median(data):
    data.sort()
    n = len(salaries)
    if n % 2 == 0:
      return (salaries[n//2 - 1] + salaries[n//2]) / 2
    else:
      return salaries[n//2]
print("Median:", calculate_median(salaries))

def calculate_mode(data):
        max_count = 0
        modes = []
        for i in data:
          count = data.count(i)
        if count > max_count:
            max_count = count
            modes = [i]
        elif count == max_count and i not in modes:
            modes.append(i)
        return modes
print("Mode:", calculate_mode(salaries))
```

```python
import seaborn as sns
import pandas as pd
df=sns.load_dataset('tips')

for col in ['total_bill','tip']:
    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Modes:", df[col].mode().tolist())

print("------------------------------------------------------")

lunch = df[df['time'] == 'Lunch']
dinner = df[df['time'] == 'Dinner']
print("LUNCH Mean bill:", lunch['total_bill'].mean())
print("LUNCH Median bill:", lunch['total_bill'].median())
print("------------------------------------------------------")
print("DINNER Mean bill:", dinner['total_bill'].mean())
print("DINNER Median bill:", dinner['total_bill'].median())
print("------------------------------------------------------")
largest_bill = df.nlargest(3, 'total_bill')
print("3 largest bills:",largest_bill[['total_bill', 'tip']])
df1 = df.drop(largest_bill.index)

# Recalculate mean and median
print("Og mean:", df['total_bill'].mean())
print("Og median:", df['total_bill'].median())

print("Mean after exclusion:", df1['total_bill'].mean())
print("Median after exclusion:", df1['total_bill'].median())
```

```python
data_1 = [12, 15, 15, 18, 20, 22, 25]
data_2 = [5, 5, 10, 10, 15, 20,0]
data_3 = [7, 14, 21, 28, 35, 0, 0]
data_4 = [100, 100, 100, 100,0,0,0]
data_5 = [1, 2, 3, 4, 5, 100,0]


def mean(data):
    total = 0
    for value in data:
        total += value

    return total / len(data)

def median(data):
    numbers = sorted(data)
    n = len(data)
    mid = n // 2

    if n % 2 == 0:
        return (data[mid - 1] + data[mid]) / 2
    else:
        return data[mid]

def mode(data):
    return max(set(data), key=data.count)

datasets = {
    "Dataset 1": data_1,
    "Dataset 2": data_2,
    "Dataset 3": data_3,
    "Dataset 4": data_4,
    "Dataset 5": data_5
}
print(f"{'Dataset':<12} {'Mean':>10} "
      f"{'Median':>10} {'Mode':>15}")
print("-" * 50)

for name, data in datasets.items():
    mean = calculate_mean(data)
    median = calculate_median(data)
    mode = calculate_mode(data)

    mode_text = str(mode) if mode else "No mode"

    print(f"{name:<12} {mean:>10.2f} "
          f"{median:>10.2f} {mode_text:>15}")
```
