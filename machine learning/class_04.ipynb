```python
import matplotlib.pyplot as plt 
import numpy as np 
from sklearn.linear_model import LinearRegression
```

```python
x = np.array([1,2,3,4,5,6,7,8]).reshape(-1,1)
y = np.array([12,22,32,42,52,62,72,80])
```

```python
model = LinearRegression()
model.fit(x,y)
```

```python
m = model.coef_
c = model.intercept_
y_pred = model.predict(x)
plt.scatter(x,y, color='red')
plt.plot(x,y_pred, color='green')   #the line is called BEST FIT LINE
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, max_error

x1 = np.linspace(12,50, num = 30, dtype = int)
x2 = np.linspace(22,60, num = 30, dtype = int)
y = np.linspace(48, 162, num=30,dtype=int)

data = {
        'x1':x1,
        'x2':x2,
        'y':y}

df=pd.DataFrame(data)

x01 = df.iloc[:,:-1]
y01=df.iloc[:,-1]
```

```python
lr = LinearRegression()
lr.fit(x01,y01)
y_pred= lr.predict(x01)
```

```python
from sklearn.metrics import mean_absolute_error
print(lr.coef_)
print(lr.intercept_)
print(mean_squared_error(y,y_pred))
print(mean_absolute_error(y,y_pred))
```
