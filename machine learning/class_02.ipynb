```python
import pandas as pd
import seaborn as sns
df= sns.load_dataset('tips')
```

```python
from sklearn.preprocessing import LabelEncoder
encoder_time = LabelEncoder()

encoded_data = encoder_time.fit_transform(df['time'])

print('classes', encoder_time.classes_)
```

```python
for category, number in zip(encoder_time.classes_, range(len(encoder_time.classes_))):
    print(category,'------------', number)
```

```python
encoder_day = LabelEncoder()

encoded_day_data = encoder_day.fit_transform(df['day'])
print('classes', encoder_day.classes_)
```

```python
for category, number in zip(encoder_day.classes_, range(len(encoder_day.classes_))):
    print(category,'------------', number)
```

```python
encoder_day.inverse_transform([0,1,2,3])
```

```python
import numpy as np
from sklearn.preprocessing import PowerTransformer
x = np.array([
    [1],
    [1],
    [3],
    [5],
    [10],
    [50]
])

pt = PowerTransformer(method = "box-cox")

x_transformed = pt.fit_transform(x)

print("original:")
print(x.ravel())

print("\nTransformed:")
print(np.round(x_transformed.ravel(),3))

print("\nLearned lambda:")
print(pt.lambdas_)
```

```python
import pandas as pd
import numpy as np

# Sample customer dataset
data = {
    'Customer_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    # Numerical columns with nulls
    'Age': [25, np.nan, 35, 42, np.nan, 29, 50, 38],
    'Annual_Income': [54000, 62000, np.nan, 85000, 71000, np.nan, 95000, 58000],
    'Credit_Score': [710, 680, 740, np.nan, 690, 720, 810, np.nan],
    # Categorical columns with nulls
    'Education': ['Bachelor', 'Master', 'Bachelor', np.nan, 'PhD', 'Master', np.nan, 'Bachelor'],
    'City': ['New York', np.nan, 'Chicago', 'Houston', np.nan, 'Miami', 'Chicago', 'New York'],
    'Employment_Status': ['Employed', 'Self-Employed', np.nan, 'Employed', 'Employed', np.nan, 'Self-Employed', 'Employed'],
    # Target / Full column
    'Purchased': ['Yes', 'No', 'No', 'Yes', 'Yes', 'No', 'Yes', 'No']
}

df = pd.DataFrame(data)

df

```

```python
from sklearn.impute import SimpleImputer

imputer_mean = SimpleImputer(strategy='mean')

imputer_mean.fit_transform(df[['Annual_Income']])
```

```python
df['new_salary']=imputer_mean.fit_transform(df[['Annual_Income']]).flatten()
df
```

```python
imputer_mode = SimpleImputer(strategy='most_frequent')
imputer_mode.fit_transform(df[['Age']])
df['new_age']=imputer_mode.fit_transform(df[['Age']]).flatten()
df
```

```python
imputer = SimpleImputer(strategy='most_frequent')
df['City']=imputer.fit_transform(df[['City']]).flatten()
df
```
