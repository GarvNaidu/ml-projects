```python
import numpy as np
```

```python
s2=pd.Series([12,13,14,15,np.nan,45,67,88,90,'pradeep'], name='series')
s2.info()
```

```python
s2.isnull() #tells null values
```

```python
s2.isnull().sum() #tells number of null values
```

```python
s2 = pd.to_numeric(s2, errors='coerce')      #converts every value to int/float, strings gets converted to NaN, so the dtype will show 'int/float'
s2
```

```python
s2=s2.fillna(1)       #fills every NaN value according to the input given
s2
```

```python
s2.value_counts()       #tells the number of times a value is repeated
```

```python
s2.unique()
```

```python
s2.nunique() #number of unique values
```

```python
s2.duplicated().sum()   
s2
```

```python
s2.where(s2>23)     #can pass to check particular conditions, and shows NaN to all the values which do not meet the conditions
```
