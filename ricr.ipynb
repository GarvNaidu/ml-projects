```python
x = input("enter a string:")
digits = 0
letters = 0
for i in x:
    if i.isdigit():
        digits+=1
    elif i.isalpha():
        letters+=1

print("letters:",letters)
print("digits:",digits)
```

```python
x = input("enter your password:")
u =0
l =0
d =0
sch =0
if len(x) >=8:
    for i in x:
        if i.isupper():
            u+=1
        elif i.islower():
            l+=1
        elif i.isdigit():
            d+=1
        elif i in '@#$%&*':
            sch +=1
    if u>=1 and l>=1 and d>=1 and sch>=1:
          print('correct password')
    else:
          print('incorrect password')
else:
    print('password length should be atleast 8 characters')
```

```python
menu = {
    'pizza':99,
    'burger':49,
    'pasta':89,
    'coffee':39,
    'salad':79,
}
#Greet
print("welcome to the restaurant!!!")
print("pizza: Rs-99\nburger: Rs-49\npasta: Rs-89\ncoffee: Rs-39\nsalad: Rs-79")

order_total = 0

item_1 = input('enter the name of item you want to order: ')
if item_1 in menu:
    order_total += menu[item_1]
    print(f'Your item has been added to your order')
else:
    print(f'Ordered item {item_1} is not available yet, SORRY!')

another_order = input('do you want to add another item? (YES/NO)')
if another_order == 'yes':
    item_2= input('enter the name of second item: ')
    if item_2 in menu:
        order_total += menu[item_2]
        print(f'item {item_2} has been added to order')
    else:
        print(f'Ordered item {item_1} is not available yet, SORRY!')

print(f'the total amt of ordered items to pay is: {order_total}')
```
