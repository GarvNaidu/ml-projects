```python
df['annualcharges']=df['monthlycharges']*12
df['annualcharges']
```

```python
def converter(x):
    if x<12:
        return 'new'
    elif 12<=x<=48:
        return 'regular'
    else:
        return 'loyal'




df['customertype']=df['tenure'].apply(converter)
df['customertype']
```
